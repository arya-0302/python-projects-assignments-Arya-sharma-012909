```python
#sum and average of first 55 natural numbers
n= int(input("Enter a number:"))
total= 0
for i in range(1, n+1):
  total +=i
average=total/n
print("sum=",total)
print("average=", average)

```

```python
#calculating factorial of 7 with a loop
n= int(input("enter a number:"))
factorial=1
for i in range(1,n+1):
  factorial*=i
print("Factorial of",n,"is",factorial)
```

```python
#take a string and reverse it
text = input("enter a string")
reverse =""
for i in text:
  reverse = i + reverse
print("reverse string:",reverse)
```

```python
#check whether input year is leap year or not
year= int(input("Enter a year"))
if year % 400 == 0:
  print("Leap year")
elif year % 100==0:
  print("Not a leap year")
elif year % 4 == 0:
  print("Leap year")
else:
  print("Not a leap year")
```
