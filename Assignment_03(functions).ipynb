```python
# 1. Write a Python program with separate functions for addition, subtraction, multiplication, and division. Take two numbers from the user and display the result.
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b


a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", addition(a, b))
print("Subtraction:", subtraction(a, b))
print("Multiplication:", multiplication(a, b))
print("Division:", division(a, b))
```

```python
# 2.Create functions to convert:Celsius → Fahrenheit, Fahrenheit → Celsius and Celsius → Kelvin.
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9

def celsius_to_kelvin(c):
    return c + 273.15


c = float(input("Enter temperature in Celsius: "))
f = float(input("Enter temperature in Fahrenheit: "))

print("Celsius to Fahrenheit:", celsius_to_fahrenheit(c))
print("Fahrenheit to Celsius:", fahrenheit_to_celsius(f))
print("Celsius to Kelvin:", celsius_to_kelvin(c))
```

```python
#3. Create a module called module_factorial.py containing functions for: Factorial Import the module into another Python program and the functions.
%%writefile module_factorial.py

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact
```

```python
#4. Write a Python function that takes a number as a parameter and checks whether the number is prime or not.
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


num = int(input("Enter a number: "))

if is_prime(num):
    print(num, "is a Prime Number")
else:
    print(num, "is Not a Prime Number")
```

```python
# 5.Write a Python function to reverse a string. (Define a function named 'string_reverse')
def string_reverse(s):
    return s[::-1]

text = input("Enter a string: ")

print("Reversed string:", string_reverse(text))
```
