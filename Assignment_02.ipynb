```python
#Python program to take an integer n as input and calculate the sum of the following series: 1^2 + 2^2 + 3^2 + ... + n^2
n = int(input("Enter a number: "))
total = 0

for i in range(1, n + 1):
    total += i ** 2

print("Sum of series =", total)
```

```python
#3.Python program to take a number as input and check whether it is a prime number using a for loop.

n = int(input("Enter a number: "))

if n <= 1:
    print("Not a prime number")
else:
    for i in range(2, n):
        if n % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```

```python
#Python program to take an integer as input and find the sum of its digits using a while loop.

n = int(input("Enter a number: "))
n = abs(n)
total = 0

while n > 0:
    digit = n % 10
    total += digit
    n = n // 10

print("Sum of digits =", total)
```

```python
#Write a Python program to take an integer n as input and reverse the number using a while loop.

n = int(input("Enter a number: "))
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

print("Reversed number =", reverse)
```

```python
# Python program to repeatedly ask the user to enter a number until the user enters 0. At the end, print the sum of all the numbers entered (excluding 0).

total = 0

while True:
    n = int(input("Enter a number: "))

    if n == 0:
        break

    total += n

print("Sum of all numbers =", total)
```

```python
#Python program to take 10 numbers from the user and store them in a list. Then find and print the largest and smallest elements without using the max() and min() functions.

numbers = []

for i in range(10):
    n = int(input("Enter a number: "))
    numbers.append(n)

largest = numbers[0]
smallest = numbers[0]

for i in numbers:
    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

print("List:", numbers)
print("Largest number:", largest)
print("Smallest number:", smallest)
```

```python
#Python program to remove all duplicate elements from a list without using set().

numbers = [10, 20, 30, 10, 40, 20, 50, 30]
unique = []

for i in numbers:
    if i not in unique:
        unique.append(i)

print("Original list:", numbers)
print("List without duplicates:", unique)
```
